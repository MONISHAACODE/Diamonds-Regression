```python
import numpy as np
from scipy import stats
z_score = np.abs(stats.zscore(df['AveragePrice']))
df_clean = df[z_score < 3]
```

```python
df_clean.describe()
```

```python
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
formula = 'AveragePrice ~ C(type) + C(region)'
model = ols(formula, df_clean).fit()
aov_table = anova_lm(model)
print(aov_table)
```

```python
aov_table['PR(>F)'] < 0.05
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df_clean['type'] = le.fit_transform(df_clean['type'])

df_clean['region'] = le.fit_transform(df_clean['region'])
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.20, random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
y_pred = lr.predict(x_train)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

print(r2_score(y_train, y_pred))
print(mean_absolute_error(y_train, y_pred))
```
